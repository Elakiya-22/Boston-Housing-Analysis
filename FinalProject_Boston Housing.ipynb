```python
import io
import pandas as pd
from js import fetch

URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/boston_housing.csv'
resp = await fetch(URL)

boston_url = io.BytesIO((await resp.arrayBuffer()).to_py())

boston_df = pd.read_csv(boston_url)

boston_df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.boxplot(x=boston_df['MEDV'])
plt.title('Boxplot of Median Value of Boston Homes (in $1000s)')
plt.show()
```

```python
boston_df.describe()
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x='CHAS', y='MEDV', data=boston_df)
plt.title('Comparison of Median House Prices by Charles River Proximity')
plt.xlabel('Near Charles River (1 = Yes, 0 = No)')
plt.ylabel('Median Value of Homes ($1000s)')
plt.show()
```

```python
boston_df.loc[(boston_df['AGE'] <= 35), 'age_group'] = '35 years and younger'
boston_df.loc[(boston_df['AGE'] > 35)&(boston_df['AGE'] < 70), 'age_group'] = 'between 35 and 70 years'
boston_df.loc[(boston_df['AGE'] >= 70), 'age_group'] = '70 years and older'

plt.figure(figsize=(10,6))
sns.boxplot(x='age_group', y='MEDV', data=boston_df)
plt.title('Median Home Value by Age Group')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='INDUS', y='NOX', data=boston_df)
plt.title('Relationship between Nitric Oxide Concentrations and Non-retail Business Acres')
plt.xlabel('Proportion of Non-retail Business Acres per Town')
plt.ylabel('Nitric Oxide Concentrations (parts per 10 million)')
plt.show()
```

```python

plt.figure(figsize=(10,6))
sns.histplot(boston_df['PTRATIO'], kde=False)
plt.title('Distribution of Pupil-Teacher Ratio by Town')
plt.xlabel('Pupil-Teacher Ratio')
plt.ylabel('Frequency')
plt.show()
```

```python
import scipy.stats
river_yes = boston_df[boston_df['CHAS'] == 1]['MEDV']
river_no = boston_df[boston_df['CHAS'] == 0]['MEDV']


t_stat, p_value = scipy.stats.ttest_ind(river_yes, river_no)

print(f"T-Statistic: {t_stat}, P-Value: {p_value}")

if p_value < 0.05:
    print("Conclusion: Since the p-value is less than 0.05, we reject the null hypothesis. There is a significant difference.")
else:
    print("Conclusion: Since the p-value is greater than 0.05, we fail to reject the null hypothesis.")
```

```python
low_age = boston_df[boston_df['age_group'] == '35 years and younger']['MEDV']
mid_age = boston_df[boston_df['age_group'] == 'between 35 and 70 years']['MEDV']
high_age = boston_df[boston_df['age_group'] == '70 years and older']['MEDV']

f_stat, p_value_anova = scipy.stats.f_oneway(low_age, mid_age, high_age)

print(f"F-Statistic: {f_stat}, P-Value: {p_value_anova}")

if p_value_anova < 0.05:
    print("Conclusion: At least one age group's mean is significantly different.")
else:
    print("Conclusion: No significant difference between age groups.")
```

```python
correlation, p_value_pearson = scipy.stats.pearsonr(boston_df['INDUS'], boston_df['NOX'])

print(f"Correlation Coefficient: {correlation}, P-Value: {p_value_pearson}")

if p_value_pearson < 0.05:
    print("Conclusion: Since the p-value is less than 0.05, we reject the null hypothesis. There is a strong positive correlation.")
else:
    print("Conclusion: There is no significant correlation.")
    
```
